```python
df = spark.read.table("data_engineering_workshop.demo1.silver_table")
```

```python
from pyspark.sql.functions import sum

kpi_total = df.agg(
    sum("Revenue").alias("total_revenue")
)
kpi_total.write.mode("overwrite").saveAsTable("data_engineering_workshop.demo1.kpi_total")

kpi_total.display()
```

```python
kpi_country = df.groupBy("Country") \
    .agg(sum("Revenue").alias("total_revenue"))
kpi_country.write.mode("overwrite").saveAsTable("data_engineering_workshop.demo1.kpi_country")
kpi_country.display()
```

```python
from pyspark.sql.functions import desc

kpi_customer = df.groupBy("Customer_Id") \
    .agg(sum("Revenue").alias("total_spent")) \
    .orderBy(desc("total_spent"))
kpi_customer.write.mode("overwrite").saveAsTable("data_engineering_workshop.demo1.kpi_customer")
kpi_customer.display()
```

```python
kpi_month = df.groupBy("Year", "Month") \
    .agg(sum("Revenue").alias("monthly_revenue")) \
    .orderBy("Year", "Month")
kpi_month.write.mode("overwrite").saveAsTable("data_engineering_workshop.demo1.kpi_month")
kpi_month.display()
```
